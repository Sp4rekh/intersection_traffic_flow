```python
import core as Graphs
import random
```

```python
graph = Graphs.Graph()
for i in range(4): 
    graph.add_node(Graphs.Vertex(str(i)))

vertices = graph.get_nodes()
print(vertices)
for i in range(4):
    graph.add_edge(Graphs.Edge(vertices[i], vertices[(i+1) % 4], directed=False))
print(graph.edges)
```

```python
graph.visualise()
```

```python
edges = graph.get_edges()
vertices = graph.get_nodes()
```

```python
for vertex in vertices:
    for edge in edges:
        if edge.contains_vertex(vertex):
            print(f"Vertex {vertex.label} is connected to edge {(edge.v1.label, edge.v2.label)}")
```
